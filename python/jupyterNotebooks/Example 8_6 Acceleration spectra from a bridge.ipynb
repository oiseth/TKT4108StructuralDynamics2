# Example 8.6: Acceleration spectra from a bridge

This example shows the analysis of a time series of accelerations on the midspan of the Hardanger bridge under excitation from wind loading.

We will estimate the cross-spectral density.



## Import the data

The 10 minutes of data is contained within a txt file. The file contains a time vector and six signals: 
$\ddot{x_1}(t), \ddot{y_1}(t), \ddot{z_1}(t), \ddot{x_2}(t), \ddot{y_2}(t), \ddot{z_2}(t)$.

The file headers use `x1`, `y1`, etc. for the corresponding acceleration signals.

It's always sensible to plot the data in the time domain to ensure it is correctly formatted.


```python
# Import and plot time series

import numpy as np
from matplotlib import pyplot as plt
from scipy import interpolate
from scipy import signal as spsi

# Import data x1, y1, and z1
data_labels=np.loadtxt('data/acc_data_hardanger.txt',delimiter=',',skiprows=0, max_rows=1, dtype=str, usecols=[0,1,2,3,4,5,6])

print('Data labels: ')
print(data_labels)

y_labels=[r'$\ddot{' + label[0] + '}_' + label[1] + '$' for label in data_labels[1:]]

t=np.loadtxt('data/acc_data_hardanger.txt',delimiter=',',skiprows=1, usecols=[0]).T # Time vector

y=np.loadtxt('data/acc_data_hardanger.txt',delimiter=',',skiprows=1, usecols=[1,2,3,4,5,6]).T # Acceleration data
 
ym=np.mean(y,axis=1)
y=y-np.atleast_2d(ym).T # Subtract the mean value

ts=t[1]-t[0]
fs=1/ts

print('Sample rate is ' + str(fs) + ' Hz')

# Plot time series
idx=-1
fig, axs = plt.subplots(2,3,figsize=(10,4))
for k1 in np.arange(2):
    for k2 in np.arange(3):
        idx=idx+1
        axs[k1,k2].plot(t,y[idx,:])
        axs[k1,k2].set_ylabel(y_labels[idx] + ' [m/s$^2$]')
        axs[k1,k2].set_xlabel('t')
        axs[k1,k2].grid(True)
        axs[k1,k2].set_xlim(0,600)
plt.tight_layout()
```

## Welch estimate of cross-spectral density

The one-sided spectral densities are expressed per Hz, with frequency $f$ in Hz. All entries of the acceleration spectral matrix have units $(\mathrm{m/s^2})^2/\mathrm{Hz}$.

Each Hann-windowed segment spans one fifth of the record (372 s). The default 50% overlap gives nine averaged segments. Zero-padding to 1024 points gives an FFT spacing of about 0.00195 Hz; the actual spectral resolution is governed by the segment duration and Hann window.

The plot below shows only the $3\times3$ block for sensor 1, with directions $x_1$, $y_1$, and $z_1$. The full $6\times6$ matrix is retained for the transformation using both sensors.


```python
# Welch estimate of cross-spectral density

import scipy

Ndivisions=5 # Segment length is one fifth of the record
Nwindow=np.ceil(len(t)/Ndivisions) # Length of window
Nfft_pow2 = 2**(np.ceil(np.log2(Nwindow))) # Next power of 2

# Spectral matrix
S_welch=np.zeros((6,6,np.int32(Nfft_pow2/2+1)),dtype=np.complex128)

# Fill spectral matrix by taking the cross spectral density between each of the 6 time series
for k1 in range(6):
    for k2 in range(6):
            f_welch, S_tmp=scipy.signal.csd(y[k1,:],y[k2,:], fs=1/ts, window='hann', nperseg=Nwindow, noverlap=None, nfft=Nfft_pow2, detrend='constant', return_onesided=True, scaling='density', axis=-1, average='mean')
            
            S_welch[k1,k2,:]=S_tmp
        
# Plot sensor 1's spectral block
sensor_labels=[r'\ddot{x}_1', r'\ddot{y}_1', r'\ddot{z}_1']
fig, axs = plt.subplots(3,3,figsize=(10,6))    
for k1 in range(3):
    for k2 in range(3):
        ax = axs[k1,k2]
        
        ax.plot(f_welch,np.real(S_welch[k1,k2,:]), label='Real') # Plot real part
        
        if k1!=k2:
            ax.plot(f_welch,np.imag(S_welch[k1,k2,:]), label='Imaginary') # Plot imaginary part (only for cross spectra)
            ax.legend(fontsize=8)
        
        ax.set_ylabel(r'$S_{' + sensor_labels[k1] + sensor_labels[k2] + '}(f)$')
        ax.set_xlabel('$f$ [Hz]')
        ax.grid(True)
        
        if k1==k2:
            ax.set_yscale('log')
            
        ax.set_xlim(0,0.5)
plt.tight_layout()


#fn_model=np.array([0.051,0.101,0.110,0.141,0.177,0.200,0.210,0.234,0.274,0.306,0.328,0.358]) # Frequencies from fe model

# Plot frequencies in spectrum
#for k1 in range(3):
 #       for n in range(len(fn_model)):
  #          axs[k1,k1].axvline(fn_model[n],color ='r',linewidth=0.5,linestyle='--')

```

## A closer look at bending and torsional modes

Many structures, bridges and buildings with regular geometries, can be analyzed using the reference figure below.

<center>
<img src="figures\Coordinate_transformation.svg">
</center>

The lower vibration modes of these structures typically fall into one of the following categories:

- *i)* a bending mode about the weak axis
- *ii)* a bending mode about the strong axis
- *iii)* a torsional mode.

Note that classifying modes of very high order or structures with more complex geometries or without clear symmetries becomes more difficult.

If we place a pair of sensors (accelerometers) at the left/right edges of the structure, we can transform their data to a reference point in the center of the bridge:

$$
\left[ \begin{array}{c}
\ddot{y}(t) \\
\ddot{z}(t) \\
\ddot{\theta}(t)
\end{array} \right] =
\underbrace{
\left[ \begin{array}{cccccc}
0 & \dfrac{1}{2} & 0 & 0 & \dfrac{1}{2} & 0 \\
0 & 0 & \dfrac{1}{2} & 0 & 0 & \dfrac{1}{2} \\
0 & 0 & -\dfrac{1}{L} & 0 & 0 & \dfrac{1}{L} \\
\end{array} \right]
}
_{\mathbf{T}}
\left[ \begin{array}{c}
\ddot{x_1}(t) \\
\ddot{y_1}(t) \\
\ddot{z_1}(t) \\
\ddot{x_2}(t) \\
\ddot{y_2}(t) \\
\ddot{z_2}(t) 
\end{array} \right]
$$

Here $L=13.7\,\mathrm{m}$ is the distance between the sensors. The transformed signals are the translational accelerations $\ddot y(t)$ and $\ddot z(t)$ in $\mathrm{m/s^2}$ and the angular acceleration $\ddot\theta(t)$ in $\mathrm{rad/s^2}$.

For a symmetric structure, these three transformed signals predominantly reflect the respective bending and torsional mode types. The transformation assumes small rotations, approximately rigid cross-sectional motion, and sensors aligned with the same coordinate axes at symmetric positions. Coupled modes and departures from these assumptions can mix the contributions. This transformation helps associate natural frequencies with mode types.

We can either transform the time series, or transform the cross-spectral density matrix of the original degrees of freedom:

$$
\mathbf{S}_m(f)= \mathbf{T} \mathbf{S}(f)  \mathbf{T}^\text{T}
$$

where $\mathbf{T}$ is the $3\times6$ transformation matrix and $\mathbf{S}_m(f)$ is a $3\times 3$ matrix for the middle point.

The translational auto- and cross-spectra have units $(\mathrm{m/s^2})^2/\mathrm{Hz}$; the angular acceleration autospectrum has units $(\mathrm{rad/s^2})^2/\mathrm{Hz}$; and the translation–rotation cross-spectra have units $(\mathrm{m/s^2})(\mathrm{rad/s^2})/\mathrm{Hz}$.


```python
L = 13.7 # Distance between sensor 1 and 2 [m]

T = np.array([
    [0, 1/2, 0, 0, 1/2, 0],
    [0, 0, 1/2, 0, 0, 1/2],
    [0, 0, -1/L, 0, 0, 1/L]
])

S_mid=np.zeros((3,3,len(f_welch)),dtype='complex')
for k in np.arange(len(f_welch)):
    S_mid[:,:,k]=T @ S_welch[:,:,k] @ T.T # Transform spectrum

m_labels=[r'\ddot{y}', r'\ddot{z}', r'\ddot{\theta}']
    
# Plot spectrum
fig, axs = plt.subplots(3,3,figsize=(10,6))    
for k1 in range(3):
    for k2 in range(3):
        ax = axs[k1,k2]
        
        ax.plot(f_welch,np.real(S_mid[k1,k2,:]), label='Real') # Plot real part
        
        if k1!=k2:
            ax.plot(f_welch,np.imag(S_mid[k1,k2,:]), label='Imaginary') # Plot imaginary part (only for cross spectra)
            ax.legend(fontsize=8)
        
        ax.set_ylabel(r'$S_{' + m_labels[k1] + m_labels[k2] + '}(f)$')
        #ax.set_ylabel(r'$S_{\theta}(f)$')
        #ax.set_ylabel(r'$S_{' + m_labels[k2] + '}(f)$')
        ax.set_xlabel(r'$f$ [Hz]')
        ax.grid(True)
        
        if k1==k2:
            ax.set_yscale('log')
            
        ax.set_xlim(0,0.5)
plt.tight_layout()

# Plot frequencies in spectrum
for k1 in range(3):
        for n in range(len(fn_model)):
            axs[k1,k1].axvline(fn_model[n],color ='r',linewidth=0.5,linestyle='--')
```
